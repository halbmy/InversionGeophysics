```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
G=np.array([[1, -1], [2, -1], [1, 1]])
d = np.array([-1, 0, 2.5])
```

```python
Ginv = np.linalg.inv(G.T @ G) @ G.T
RM = Ginv@G
RD = G@Ginv
fig, ax = plt.subplots(ncols=2)
ax[0].imshow(RM, vmin=-1, vmax=1, cmap="bwr")
im = ax[1].imshow(RD, vmin=-1, vmax=1, cmap="bwr")
plt.colorbar(im)
```

```python
im = plt.imshow(RD, vmin=-1, vmax=1, cmap="bwr")
plt.colorbar(im)
```

```python
mLS = Ginv @ d
# mLS, *_ = np.linalg.lstsq(G, d, rcond=None)
print(mLS)
```

```python
fig, ax = plt.subplots()
x = np.array([0.65, 1.1])
dd = 0.081
for i in range(len(G)):
    y = (d[i] - G[i, 0] * x) / G[i, 1]
    dy = dd/G[i, 1]
    ax.fill_between(x, y-dy, y+dy, color=f"C{i}", alpha=0.5)
    ax.plot(x, y, "-", label=f"{i}", color=f"C{i}")

ax.plot(mLS[0], mLS[1], "r*", label="m-LS")
ax.set_aspect(1.0)
ax.set_ylim(1.6, 2.0)
ax.legend()
ax.grid()
```

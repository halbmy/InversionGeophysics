```python
import numpy as np
from matplotlib import pyplot as plt
import pygimli as pg
from pygimli.physics.ert import VESModelling
from objfun import showPhiD
```

```python
x = np.linspace(0, 10, 21)
a, b = 0.5, 0.5
y = a*x + b
error = 0.5
y += np.random.randn(len(y)) * error
plt.plot(x, y, "x")
plt.plot(x, a*x+b, "-")
```

```python
aa = np.linspace(0, 1, 31)
bb = np.linspace(-1, 2, 31)
CHI2 = np.zeros([len(aa), len(bb)])
for i, ai in enumerate(aa):
    for j, bi in enumerate(bb):
        yy = x*ai + bi
        CHI2[i, j] = np.mean(((yy-y)/error)**2)
```

```python
showPhiD(bb, aa, CHI2, vmax=10);
```
